**T1.** A product costs ₹1,499 before GST. GST is 18%. Calculate the GST amount and the final price, rounded to 2 decimals.

```python
price = 1499
gst_rate = 0.18
gst = price * gst_rate
final_price = price + gst
print(round(gst, 2), round(final_price, 2))
```

**T3.** Daily orders for a week are [120, 135, 98, 160, 175, 210, 190]. Find the total, average (1 decimal), highest and lowest.

```python
daily_orders = [120, 135, 98, 160, 175, 210, 190]
print(sum(daily_orders), round(sum(daily_orders) / len(daily_orders), 1), max(daily_orders), min(daily_orders))
```

**T5.** City sales (₹ lakh): {'Mumbai': 42.5, 'Delhi': 38.2, 'Bengaluru': 45.1, 'Pune': 18.7}. Find the total and the top city.

```python
city_sales = {'Mumbai': 42.5, 'Delhi': 38.2, 'Bengaluru': 45.1, 'Pune': 18.7}
total = sum(city_sales.values())
top_city = max(city_sales, key=city_sales.get)
print(round(total, 1), top_city)
```

**T6.** Loop over city_sales and print each city that is above the target of ₹40 lakh.

```python
target = 40
for city, sales in city_sales.items():
    if sales > target:
        print(city, 'is above target:', sales)
```

**T7.** Write a function discount(amount) that returns 10% off for orders ≥ ₹5,000, 5% for ≥ ₹2,000, else 0. Test with 6500, 2500 and 999.

```python
def discount(amount):
    if amount >= 5000:
        return amount * 0.10
    elif amount >= 2000:
        return amount * 0.05
    else:
        return 0

print(discount(6500), discount(2500), discount(999))
```

**T8.** Write delivery_status(days_taken, promised_days=3) returning 'On time' or 'Late'. Test with 2, 5 and (5, 6).

```python
def delivery_status(days_taken, promised_days=3):
    return 'On time' if days_taken <= promised_days else 'Late'

print(delivery_status(2), delivery_status(5), delivery_status(5, 6))
```

**T9.** Order values are [450, 2300, 1299, 5600, 899]. Use list comprehensions to (a) add 18% GST to each, rounded, and (b) keep only orders above ₹1,000.

```python
values = [450, 2300, 1299, 5600, 899]
with_gst = [round(v * 1.18) for v in values]
big = [v for v in values if v > 1000]
print(with_gst)
print(big)
```

**T10.** Count payment modes in ['UPI', 'Card', 'UPI', 'COD', 'UPI', 'Card', 'COD', 'UPI'] using a dictionary.

```python
payments = ['UPI', 'Card', 'UPI', 'COD', 'UPI', 'Card', 'COD', 'UPI']
counts = {}
for p in payments:
    counts[p] = counts.get(p, 0) + 1
print(counts)
```

**T11.** Using datetime, find the days between order date 2026-09-25 and delivery date 2026-09-29, and the weekday of the delivery.

```python
from datetime import date
order_date = date(2026, 9, 25)
delivery_date = date(2026, 9, 29)
print((delivery_date - order_date).days, delivery_date.strftime('%A'))
```

**T12.** Mini project: from the list of order dictionaries, calculate revenue by city, the top city and the average order value.

```python
orders = [
    {'id': 1, 'city': 'Pune', 'amount': 1200},
    {'id': 2, 'city': 'Delhi', 'amount': 3400},
    {'id': 3, 'city': 'Pune', 'amount': 800},
    {'id': 4, 'city': 'Mumbai', 'amount': 2500},
    {'id': 5, 'city': 'Delhi', 'amount': 1600},
]
revenue = {}
for o in orders:
    revenue[o['city']] = revenue.get(o['city'], 0) + o['amount']
aov = sum(o['amount'] for o in orders) / len(orders)
print(revenue)
print(max(revenue, key=revenue.get), aov)
```
